```python
import pandas as pd

# Construct the special URL for CSV export
sheet_id = '1YPDqxJ6Xi3pyGivlEy-MImJHJ1FKxgNQ'
gid = '49811916'
url = f'https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=csv&gid={gid}'

# Read the data into a pandas DataFrame
df = pd.read_csv(url)

# Display the first few rows to confirm it worked
print(df.head())
```

```python
# Convert datatypes
df['Order ID'] = df['Order ID'].astype(int)
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
df['Price'] = df['Price'].astype(float)
df['Quantity'] = df['Quantity'].astype(float)

# Create Revenue column
df['Revenue'] = df['Price'] * df['Quantity']

print(df.info())
print(df.head())

```

```python
# Strip spaces in column names
df.columns = df.columns.str.strip()

# Drop empty/unnecessary columns like 'Unnamed'
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

# Drop missing values
df = df.dropna()

# Convert datatypes
df['Order ID'] = df['Order ID'].astype(int)
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
df['Price'] = df['Price'].astype(float)
df['Quantity'] = df['Quantity'].astype(float)

# Create Revenue column
df['Revenue'] = df['Price'] * df['Quantity']

print("Dataset after cleaning:")
print(df.info())
print(df.head())
```

```python
# Shape
print(f"The dataset has {df.shape[0]} rows and {df.shape[1]} columns.\n")

# Descriptive statistics
print("Descriptive Statistics:")
print(df.describe())

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
#Total Revenue
total_revenue = df['Revenue'].sum()
print(f"Q1: Total Revenue = €{total_revenue:,.2f}")
```

```python
# Q1: Most Preferred Payment Method
preferred_payment = df['Payment Method'].value_counts().idxmax()
print(f"Q1: Most Preferred Payment Method = {preferred_payment}")

# Q2: Most Selling Product by Quantity and by Revenue
top_product_qty = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False).head(1)
top_product_rev = df.groupby('Product')['Revenue'].sum().sort_values(ascending=False).head(1)
print("\nQ2: Most Selling Product by Quantity:")
print(top_product_qty)
print("\nQ2: Most Selling Product by Revenue:")
print(top_product_rev)

# Q3: City with Maximum Revenue & Manager with Maximum Revenue
city_rev = df.groupby('City')['Revenue'].sum().sort_values(ascending=False).head(1)
manager_rev = df.groupby('Manager')['Revenue'].sum().sort_values(ascending=False).head(1)
print("\nQ3: City with Maximum Revenue:")
print(city_rev)
print("\nQ3: Manager with Maximum Revenue:")
print(manager_rev)

# Q4: Average Revenue
avg_rev = df['Revenue'].mean()
print(f"\nQ4: Average Revenue = €{avg_rev:,.2f}")

# Q5: Average Revenue of November & December
df['Month'] = df['Date'].dt.month
nov_dec_rev = df[df['Month'].isin([11,12])]['Revenue'].mean()
print(f"\nQ5: Average Revenue (Nov & Dec) = €{nov_dec_rev:,.2f}")

# Q6: Standard Deviation of Revenue and Quantity
std_rev = df['Revenue'].std()
std_qty = df['Quantity'].std()
print(f"\nQ6: Std Deviation -> Revenue: {std_rev:,.2f}, Quantity: {std_qty:,.2f}")

# Q7: Variance of Revenue and Quantity
var_rev = df['Revenue'].var()
var_qty = df['Quantity'].var()
print(f"\nQ7: Variance -> Revenue: {var_rev:,.2f}, Quantity: {var_qty:,.2f}")

# Q8: Revenue trend over time (increasing or decreasing)
daily_rev = df.groupby('Date')['Revenue'].sum()
trend = "increasing 📈" if daily_rev.iloc[-1] > daily_rev.iloc[0] else "decreasing 📉"
print(f"\nQ8: Revenue trend over time is {trend}")

# Q9: Average Quantity & Revenue for each Product
avg_prod = df.groupby('Product')[['Quantity','Revenue']].mean()
print("\nQ9: Average Quantity & Revenue per Product:")
print(avg_prod.head(10))  # first 10 products

# Q10: Total Number of Orders/Sales
total_orders = df['Order ID'].nunique()
print(f"\nQ10: Total Number of Orders = {total_orders}")
```

```python
# BUSINESS ANALYSIS Q1 - Q10 (with charts)
# =========================

import matplotlib.pyplot as plt
import seaborn as sns

# Q1: Most Preferred Payment Method
preferred_payment_counts = df['Payment Method'].value_counts()
preferred_payment = preferred_payment_counts.idxmax()
print(f"Q1: Most Preferred Payment Method = {preferred_payment}")

plt.figure(figsize=(6,4))
sns.barplot(x=preferred_payment_counts.index, y=preferred_payment_counts.values, palette="Blues")
plt.title("Most Preferred Payment Method")
plt.ylabel("Count of Transactions")
plt.xlabel("Payment Method")
plt.show()


# Q2: Most Selling Product by Quantity and by Revenue
top_product_qty = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
top_product_rev = df.groupby('Product')['Revenue'].sum().sort_values(ascending=False)

print("\nQ2: Most Selling Product by Quantity (Top 1):")
print(top_product_qty.head(1))
print("\nQ2: Most Selling Product by Revenue (Top 1):")
print(top_product_rev.head(1))

plt.figure(figsize=(10,4))
sns.barplot(x=top_product_qty.head(10).index, y=top_product_qty.head(10).values, palette="Greens")
plt.title("Top 10 Products by Quantity Sold")
plt.xticks(rotation=45)
plt.ylabel("Quantity Sold")
plt.show()

plt.figure(figsize=(10,4))
sns.barplot(x=top_product_rev.head(10).index, y=top_product_rev.head(10).values, palette="Oranges")
plt.title("Top 10 Products by Revenue")
plt.xticks(rotation=45)
plt.ylabel("Revenue (€)")
plt.show()


# Q3: City with Maximum Revenue & Manager with Maximum Revenue
city_rev = df.groupby('City')['Revenue'].sum().sort_values(ascending=False)
manager_rev = df.groupby('Manager')['Revenue'].sum().sort_values(ascending=False)

print("\nQ3: City with Maximum Revenue:")
print(city_rev.head(1))
print("\nQ3: Manager with Maximum Revenue:")
print(manager_rev.head(1))

plt.figure(figsize=(8,4))
sns.barplot(x=city_rev.head(10).index, y=city_rev.head(10).values, palette="Purples")
plt.title("Top 10 Cities by Revenue")
plt.xticks(rotation=45)
plt.ylabel("Revenue (€)")
plt.show()


# Q4: Average Revenue
avg_rev = df['Revenue'].mean()
print(f"\nQ4: Average Revenue = €{avg_rev:,.2f}")


# Q5: Average Revenue of November & December
df['Month'] = df['Date'].dt.month
nov_dec_rev = df[df['Month'].isin([11,12])]['Revenue'].mean()
print(f"\nQ5: Average Revenue (Nov & Dec) = €{nov_dec_rev:,.2f}")


# Q6: Standard Deviation of Revenue and Quantity
std_rev = df['Revenue'].std()
std_qty = df['Quantity'].std()
print(f"\nQ6: Std Deviation -> Revenue: {std_rev:,.2f}, Quantity: {std_qty:,.2f}")


# Q7: Variance of Revenue and Quantity
var_rev = df['Revenue'].var()
var_qty = df['Quantity'].var()
print(f"\nQ7: Variance -> Revenue: {var_rev:,.2f}, Quantity: {var_qty:,.2f}")


# Q8: Revenue trend over time (increasing or decreasing)
daily_rev = df.groupby('Date')['Revenue'].sum()
trend = "increasing 📈" if daily_rev.iloc[-1] > daily_rev.iloc[0] else "decreasing 📉"
print(f"\nQ8: Revenue trend over time is {trend}")

plt.figure(figsize=(10,5))
plt.plot(daily_rev.index, daily_rev.values, marker='o', color="blue")
plt.title("Daily Revenue Trend")
plt.ylabel("Revenue (€)")
plt.xlabel("Date")
plt.xticks(rotation=45)
plt.show()


# Q9: Average Quantity & Revenue for each Product
avg_prod = df.groupby('Product')[['Quantity','Revenue']].mean()
print("\nQ9: Average Quantity & Revenue per Product (first 10 shown):")
print(avg_prod.head(10))

plt.figure(figsize=(10,5))
sns.scatterplot(data=avg_prod, x="Quantity", y="Revenue")
plt.title("Avg Quantity vs Avg Revenue per Product")
plt.xlabel("Average Quantity")
plt.ylabel("Average Revenue (€)")
plt.show()


# Q10: Total Number of Orders/Sales
total_orders = df['Order ID'].nunique()
print(f"\nQ10: Total Number of Orders = {total_orders}")
```
